```python
expenses=[]

def entry():
    category=input("Enter category")
    price=int(input("Enter price"))
    expense=[category,price]
    expenses.append(expense)

def view():
    print(expenses)
    

def totalexpense():
    total=0
    for expense in expenses:
        total= total+expense[1]
    print(total)
        
def highestExpense():
    high=0
    for expense in expenses:
        if expense[1] > high:
             high = expense[1]
    print(high)

while True:
    print("======Expense Tracker========")
    print("Enter 1 to make entry of Expenses")
    choice=int(input("Enter choice to perform Operation"))
    match(choice):
        case 1:
            entry()
        case 2:
            view()
        case 3:
            totalexpense()
        case 4:
            highestExpense()
            break
            
```
