```python
from pathlib import Path
import os
def fileList():
    path=Path('')
    items=list(path.rglob('*'))
    for i,items in enumerate(items):
        print(f"{i+1} = {items}")
    
def menu():
    print("Press 1 for creating a file")
    print("Press 2 for reading a  file")
    print("Press 3 to update a file")
    print("Press 4 to delete a file")
    check=int(input("Enter choice"))
    if check == 1:
        createFile()
    if check == 2:
        readFile()
    if check == 3:
        updateFile()
    if check == 4:
        deleteFile()
def createFile():
    try:
        fileList()
        name=input("Please tell your file name")
        p=Path(name)
        if not p.exists():
            with open(p,"w") as fs:
                data=input("What you want to write")
                fs.write(data)
            print("File created successfully")
        else:
            print("This file already exists")
    except Exception as e:
        print(f"Error occured as {e}")
def readFile():
    try:
        fileList()
        name=input("Which file you want to read")
        p=Path(name)
        if p.exists() and p.is_file():
            with open(p,"r") as fs:
                data=fs.read()
                print(data)
            print("File read Successfully")
        else:
            print("File does not exists")
    except Exception as e:
        print(e)
def updateFile():
    try:
        fileList()
        name=input("Which file you want to update")
        p=Path(name)
        if p.exists() and p.is_file():
            print("Press 1 for changing the name of file")
            print("Press 2 for overwrite the file")
            print("Press 3 for appending data")
            res=int(input("Tell your response"))
            if res == 1:
                name2=input("Tell your new file name")
                p2=Path(name2)
                p.rename(p2)
            print("File Updated Successfully")
            if res == 2:
                with open(p,"w") as fs:
                    data=input("Tell what you want to write")
                    fs.write(data)
            print("Written Successfully")
            if res == 3:
                with open(p,"a") as fs:
                    data=input("Tell what you want to write")
                    fs.write(""+data) 
            print("Data Appended Successfully")
    except Exception as e:
        print(e)
def deleteFile():
    try:
        fileList()
        name=input("Which file you want to delete")
        p=Path(name)
        if p.exists() and p.is_file():
            os.remove(p)
            print("File removed successfullt")
        else:
            print("No such file exists")
    except Exception as e:
        print(e)
menu()
```
