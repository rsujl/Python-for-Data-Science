### 07) WAP Calculate Area and Circumfrence of Circle

```python
PI=3.14
radius=float(input("Enter Radius"))
areaofCricle=PI*radius*radius
print("Area is :" ,areaofCricle)
perimeterofCricle=(2*PI*radius)
print("Perimeter is:" ,perimeterofCricle)
```

### 10) WAP to convert degree to Fahrenheit and vice versa.

```python
c=float(input("Enter Celsius"))
fah=((c*9/5)+32)
print(fah)
f = float(input("Enter Fahrenheit: "))
c = (f - 32) * 5 / 9
print("Celsius =", c)
```

### 12) Compute distance between two points taking input from the user (Pythagorean Theorem).

```python
x1=float(input("Enter x1 distance"))
y1=float(input("Enter y1 distance"))
x2=float(input("Enter x2 distance"))
y2=float(input("Enter y2 distance"))
distance=((x2-x1)**2 + (y2-y1)**2)**5
print(distance)
```

### 13) WAP to convert seconds into hours, minutes & seconds and print in HH:MM:SS
[e.g. 10000 seconds mean 2:46:40 (2 Hours, 46 Minutes, 40Seconds)]

```python
seconds=int(input("Enter Seconds"))
hour=(seconds//3600)
minutes=(seconds%3600)//60
seconds=seconds%60
print(hour,minutes,seconds)
```

### 14) WAP to enter distance into kilometer and convert it into meter, feet,inches, and centimeter

```python
km = float(input("Enter distance in kilometer: "))

meter = km * 1000
feet = km * 3280.84
inches = km * 39370.1
centimeter = km * 100000

print("Meter =", meter)
print("Feet =", feet)
print("Inches =", inches)
print("Centimeter =", centimeter)
```
