```python
users = []

def register():
    name = input("Enter name: ").strip().lower()
    gmail = input("Enter Gmail: ")
    password = input("Enter password: ")

    user = {
        "name": name,
        "gmail": gmail,
        "password": password
    }

    users.append(user)

    print("Registration successful")


def login():
    gmail = input("Enter Gmail: ")
    password = input("Enter password: ")

    for user in users:
        if user["gmail"] == gmail and user["password"] == password:
            print("Logged in")
            print(f"Name is {user['name']}")
            print(f"Gmail is {user['gmail']}")
            return

        print("Invalid Login")
A=login

def main():
    register()
    A()
def play():
    main()
play()
```
