```python
import random
computer=random.randint(1,100)
print(computer)
ctn=0
while ctn<3:
    guess=int(input("Enter Number to guess in 3 Attempts"))
    if(guess<computer):
        print("Number is smaller guess larger")
    elif (guess>computer):
        print("Number is larger guess smaller")
    elif(guess == computer):
        print("Guess Correctly")
        break
    ctn+=1
else:
        print("3 Attempts is Used ")
```

```python
details = []

def insert():
    bookName = input("Enter book name")
    bookId = int(input("Enter book id"))

    detail = {
        "bookName": bookName,
        "bookId": bookId
    }

    details.append(detail)
    
def view():
    print(details)

def edit():
    bookId=int(input("Enter book Id to find and then edit"))
    found=False

    for detail in details:
        if detail["bookId"]==bookId:
            print("Book found")
            found=True
            newbookName=input("Enter new Book name")
            detail["bookName"]=newbookName
            print("Book Name updated Successfully")
    if found==False:
        print("Book not found id is different")
        
def delete():
    bookId = int(input("Enter book Id to delete: "))

    for detail in details:
        if detail["bookId"] == bookId:
            print("Book found")
            details.remove(detail)
            print("Book deleted successfully")
            
    
while True:
    print("Enter 1 to enter book details")
    print("Enter 2 to view details")
    print("Enter 3 to edit details")
    print("Enter 4 to delete details")
    print("Enter 0 to stop")
    choice=int(input("Enter choice to continue"))

    match(choice):
        case 1:
            insert()
        case 2:
            view()
        case 3:
            edit()
        case 4:
            delete()
        case 0:
            print("Inavlid Input or Case !!!!!!!!!!!!")
            break

```

```python
n=int(input("Enter Number"))
def prime(n):
    if n<2:
        print("Not a prime number")
    else:
        for i in range(2,n):
            if n%i==0:
                print("Not a prime Number")
                break
        else:
                print(f"Number {n} is Prime")
prime(n)
```
