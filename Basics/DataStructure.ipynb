### Tuple

A tuple is similar to list  . The difference is that we cannot change the element of tuple once assigned

In short Tuple is :
1.Immutable
2.cannot change once created

characteristics:
1.Ordered
2.Unchangable
3.Allows Duplicate



```python
tup=(1,2,3)
tup
```

```python
tup=()
tup
```

```python
tup=(1,)
type(tup)
```

```python
t1=(1,2,3,[1,2,3])
t1
```

```python
t5=(1,2,3,(4,5))
t5
```

```python
t6=tuple([1,2,3,(7,8),True])
t6
```

### Accessing and Indexing in tuple

```python
t6=tuple([1,2,3,(7,8),True])
t6[1]
```

```python
t6=tuple([1,2,3,(7,8),True])
t6[0:4:4]
t6[-1]
```
