```python
numlst=list()
while True:
    inp=input("Enter number")
    if inp.lower() == 'done':
        break
    else:
        value=int(inp)
        numlst.append(value)
print(numlst)
```

```python
list=[]

def insert():
    n=int(input("Enter size of an arrray"))
    for i in range(n):
        inp=input("Enter elements")
        if inp.lower()=="done":
            break
        else:
            value=int(inp)
            list.append(value)
insert()
print(list)        
```

```python
elements_list=[]
def ques():
    while True:
        inp=input("Enter Elements")
        if inp.lower()=="done":
            break
        else:
            elements_list.append(int(inp))
ques()
print(elements_list)
for i in elements_list:
    if i>=0:
        print(f"{i} is positive")
    else:
        print(f"{i} is negative")
sum=0
for i in elements_list:
    sum+=i
    ans=sum/len(elements_list)
print(f"Sum is = {sum} ")
print(f"Mean is ={ans}")
largest=elements_list[0]
for i in elements_list:
    if i>largest:
        largest=i
print(f"Largest number is = {largest}")
elements_list.sort()
second_largest=elements_list[-2]
print(f"Second Largest is {second_largest}")
a=sorted(elements_list)
print(a)
if elements_list==a:
    print("List is Sorted")
else:
    print("List is not sorted")
```
