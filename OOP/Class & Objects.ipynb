```python
class Atm:
     def __init__(self):
         print(id(self))
         self.pin=''
         self.balance=0
     def menu(self):
        user_input=input(
            f"hi how can i help you \n"
            f"Press 1 to create Pin \n"
            f"Press 2 to change pin \n"
            f"Press 3 to check balance \n"
            f"Press 4 to to withdraw \n"
            f"Anyhting else to Exit n"
        )
        if user_input == "1":
            self.createPin()
        elif user_input == "2":
            self.changePin()
        elif user_input == "3":
            self.checkBalance()
        elif user_input == "4":
            self.withdrawBalance()
        else:
            exit()
```

```python
def createPin(self):
    user_pin=input("Enter Pin")
    self.pin=user_pin
    user_balance=int(input("Enter balance"))
    self.balance=user_balance
    print("Pin created Successfully")
    print(self.pin)
    print(self.balance)
    self.menu()
```

```python
Atm.createPin=createPin
```

```python
def changePin(self):
    old_pin=input("Enter Old pin")
    if old_pin == self.pin:
        newPin=input("Enter New Pin")
        self.pin=newPin
        print("New Pin is",newPin)
        self.menu()
```

```python
Atm.changePin=changePin
```

```python
def checkBalance(self):
    user_pin=input("Enter Pin")
    if user_pin==self.pin:
        print("Balance is",self.balance)
        self.menu()
    else:
        print("Invalid Pin")
        self.menu()
```

```python
Atm.checkBalance=checkBalance
```

```python
def withdrawBalance(self):
    user_pin=input("Enter Pin")
    if user_pin==self.pin:
        useramt=input("Enter amount to withdraw")
        self.balance=int(self.balance)-int(useramt)
        print("Amount Debited",useramt)
        print("Remaining Amount",self.balance)
    else:
        print("Invalid Pin")
```

```python
Atm.withdrawBalance=withdrawBalance
```
