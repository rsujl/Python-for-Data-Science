```python
class Atm:
     def __init__(self):
         print(id(self))
         self.pin=''
         self.balance=0
     def menu(self):
        user_input=input(
            f"hi how can i help you \n"
            f"Press 1 to create Pin \n"
            f"Press 2 to change pin \n"
            f"Press 3 to check balance \n"
            f"Press 4 to to withdraw \n"
            f"Anyhting else to Exit n"
        )
        if user_input == "1":
            self.createPin()
        elif user_input == "2":
            self.changePin()
        elif user_input == "3":
            self.checkBalance()
        elif user_input == "4":
            self.withdrawBalance()
        else:
            exit()
```

```python
def createPin(self):
    user_pin=input("Enter Pin")
    self.pin=user_pin
    user_balance=int(input("Enter balance"))
    self.balance=user_balance
    print("Pin created Successfully")
    print(self.pin)
    print(self.balance)
    self.menu()
```

```python
Atm.createPin=createPin
```

```python
def changePin(self):
    old_pin=input("Enter Old pin")
    if old_pin == self.pin:
        newPin=input("Enter New Pin")
        self.pin=newPin
        print("New Pin is",newPin)
        self.menu()
```

```python
Atm.changePin=changePin
```

```python
def checkBalance(self):
    user_pin=input("Enter Pin")
    if user_pin==self.pin:
        print("Balance is",self.balance)
        self.menu()
    else:
        print("Invalid Pin")
        self.menu()
```

```python
Atm.checkBalance=checkBalance
```

```python
def withdrawBalance(self):
    user_pin=input("Enter Pin")
    if user_pin==self.pin:
        useramt=input("Enter amount to withdraw")
        self.balance=int(self.balance)-int(useramt)
        print("Amount Debited",useramt)
        print("Remaining Amount",self.balance)
    else:
        print("Invalid Pin")
```

```python
Atm.withdrawBalance=withdrawBalance
```

### I see exactly what is confusing you! You are looking at print(s1.total) and thinking: "Wait, total is a class variable, and we are accessing it through an instance s1. Why does that work? Shouldn't it fail?"
In Python, you can access class variables (and class methods) using both the class name (Student.total) and an object instance (s1.total).
Here is exactly how Python handles this behind the scenes:
1. Lookup Priority (The Scope Chain)
When you type s1.total, Python looks for the variable in a specific order:
1. Instance Check: It first looks inside the s1 object's individual dictionary (s1.__dict__) to see if total belongs unique to Sujal. It doesn't find it there.
2. Class Fallback: Because it didn't find it in the instance, Python automatically looks "upstairs" at the class level (Student). It finds total = 1 there and successfully prints it.
2. The Difference Between Accessing and Modifying
While you can read a class variable using an instance, modifying it behaves very differently:
python
s1 = Student("Sujal")
s2 = Student("Amit")

# ❌ BAD: Trying to change class state using an instance
s1.clg_name = "IIT" 
print(s1.clg_name)  # Output: "IIT" (Created a NEW variable unique only to s1)
print(s2.clg_name)  # Output: "LD College" (s2 never saw the change)

#  GOOD: Changing class state using the Class or a Class Method
Student.clg_name = "IIT"
print(s1.clg_name)  # Output: "IIT"
print(s2.clg_name)  # Output: "IIT" (Both updated successfully!)
Use code with caution.
3. Calling Class Methods via Instances
The exact same rule applies to your class method s1.passing(). Even though passing is a @classmethod, Python allows you to call it using the instance s1. Under the hood, Python automatically bypasses the instance and passes the class (Student) into the cls parameter anyway.
