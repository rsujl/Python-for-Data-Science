```python
try:
    print("Sujal")
except Exception as e:
    print(e)
else:
    print("Done")
finally:
    print("This wil executed finally")
```

```python
try:
    name=input("Enter name")
    age=int(input("Enter age"))
    if not name.isalpha():
        raise ValueError("Name must contain letters only!")
except Exception as e:
    print(e)
else:
    print(f"Name is {name} and age is {age}")
finally:
    print("Registration done Successfully")
```
